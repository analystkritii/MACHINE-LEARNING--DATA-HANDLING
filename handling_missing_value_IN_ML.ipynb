```python
import pandas as pd
import numpy as np
#created the dataset
#data collection
df_retail = pd.DataFrame({
'Customer_ID': [1,2,3,4,5,6],
'Age': [25,30,28,32,29,31],
'Annual_Spend': [50000, np.nan, 48000, np.nan, 51000, 49500]
})
df_retail
```

```python
df_retail.isnull().sum()
```

```python
len(df_retail)
```

```python
(df_retail.isnull().sum()/len(df_retail))*100
```

```python
df_retail['Annual_Spend'].skew()
```

```python
df_retail['Annual_Spend']=df_retail['Annual_Spend'].fillna(
df_retail['Annual_Spend'].mean(),
inplace=True
)
```

```python
df_retail
```

```python
df_salary = pd.DataFrame({
'Employee': ['A','B','C','D','E','F'],
'Salary': [30000, 35000, 40000, 500000, 42000, np.nan]
})
df_salary
```

```python
(df_salary.isnull().sum()/len(df_salary))*100
```

```python
df_salary['Salary'].skew()
```

```python
df_salary['Salary'].fillna(
df_salary['Salary'].median(),
inplace=True
)
```

```python
df_salary
```

```python
df_customer = pd.DataFrame({
'Customer_ID': [1,2,3,4,5,6],
'City': ['Mumbai','Delhi',np.nan,'Mumbai',np.nan,'Delhi']
})
df_customer
```

```python
(df_customer.isnull().sum()/len(df_customer))*100
```

```python
df_customer['City'].fillna(
df_customer['City'].mode()[0],
inplace=True
)
```

```python
df_customer
```

```python
df_customer['City'].fillna("Unknown", inplace=True)
```
